## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
import pandas as pd
from pathlib import Path

csv_path = Path("../../data/raw/content_refresh_anonymized.csv")
if not csv_path.exists():
    raise FileNotFoundError(f"Could not find {csv_path}")

df = pd.read_csv(csv_path)
lane_df = df[[
    "content_id",
    "client_id",
    "impressions_90d",
    "sessions_90d",
    "trend_direction",
    "ctr",
    "avg_position",
]].copy()

print(f"Rows in starter slice: {len(lane_df):,}")
print("One row = one content page")
lane_df.head()
```

```python
lane_df["priority_proxy"] = (
    (lane_df["trend_direction"] == "down").astype(int)
    + (lane_df["impressions_90d"] > 500).astype(int)
    + (lane_df["avg_position"] < 10).astype(int)
)

lane_df["priority_proxy"] = lane_df["priority_proxy"].clip(upper=3)

lane_df[["content_id", "trend_direction", "priority_proxy"]].head(10)
```

one row = one page 
the dataframe row represents one content item with it's signals and metrics

## 5. Why ML beats a fixed rule here
